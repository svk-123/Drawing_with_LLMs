````python
import re
import json

def extract_json_from_text(text):
    """
    Extracts the first JSON object found in a string and returns it as a Python dict.
    If no valid JSON object is found, returns None.
    """
    try:
        # Use regex to extract content between ```json ... ```
        match = re.search(r"```json\s*(\{.*?\})\s*```", text, re.DOTALL)
        if match:
            json_str = match.group(1)
            json_str = json_str.replace("'", '"')

            return json.loads(json_str)
        else:
            print("No JSON block found.")
            return 0
    except json.JSONDecodeError as e:
        print(f"JSON decode error: {e}")
        return 0

````

````python
df_vqa['parsed_json'] = df_vqa['response'].apply(extract_json_from_text)
````
