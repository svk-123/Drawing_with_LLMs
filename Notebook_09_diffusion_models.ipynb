```python
from PIL import Image
from starvector.model.starvector_arch import StarVectorForCausalLM
from starvector.data.util import process_and_rasterize_svg
model_name = "starvector/starvector-1b-im2svg"
starvector = StarVectorForCausalLM.from_pretrained(model_name,torch_dtype=torch.float16)
starvector.cuda()
starvector.eval()

def get_svg(starvector):
    image_pil = Image.open('tmp.png')
    image = starvector.process_images([image_pil])[0].cuda()
    batch = {"image": image}
    
    raw_svg = starvector.generate_im2svg(batch, max_length=1000)[0]
    svg, _ = process_and_rasterize_svg(raw_svg)

    return svg, raw_svg
```

```python
def get_image_svg(pipe,starvector,prompt):

    # Set generation parameters
    height = 256  # Image height in pixels
    width = 256  # Image width in pixels
    num_inference_steps = 50  # Number of denoising steps
    guidance_scale = 10  # Scale for classifier-free guidance
    num_images_per_prompt = 1  # Number of images to generate per prompt
    eta = 0.0  # Corresponds to η in the DDIM paper
    
    # Generate the image
    image = pipe(prompt=prompt, height=height, width=width, num_inference_steps=num_inference_steps,
                 guidance_scale=guidance_scale, num_images_per_prompt=num_images_per_prompt, eta=eta).images[0]
    
    # Save the image
    image.save("tmp.png")

    #call svg model to get code
    svg,raw_svg=get_svg(starvector)

    return svg,raw_svg
```

```python
svg,raw_svg=get_image_svg(pipe,starvector,'sun rising in the east beach')
```
